<h2 style="color:#00796b;">🤯 Operaciones lógicas: decisiones con conejomagia</h2>

<p>Ahora que ya conoces las variables <code>True</code> y <code>False</code>, es momento de combinarlas. ¡Como hacer un hechizo con más de un ingrediente!</p>

<h3>🐰 1. <code>and</code> — "Y además..."</h3>
<p>Solo devuelve <code>True</code> si <strong>todo es verdadero</strong>.</p>

<pre style="background:#c8e6c9; padding:0.8em; border-radius:6px;">
tiene_varita = True
sabe_hechizo = True
print(tiene_varita and sabe_hechizo)   # ✅ True
</pre>

<p>Pero si uno falla...</p>

<pre style="background:#ffcdd2; padding:0.8em; border-radius:6px;">
tiene_varita = True
sabe_hechizo = False
print(tiene_varita and sabe_hechizo)   # ❌ False
</pre>

<h3>🦄 2. <code>or</code> — "Con que uno funcione..."</h3>
<p>Devuelve <code>True</code> si <strong>al menos uno</strong> es verdadero.</p>

<pre style="background:#fff9c4; padding:0.8em; border-radius:6px;">
tiene_capa = False
tiene_escoba = True
print(tiene_capa or tiene_escoba)   # ✅ True
</pre>

<h3>🙃 3. <code>not</code> — "Lo contrario, por favor"</h3>
<p>Cambia <code>True</code> a <code>False</code> y viceversa. ¡Es un hechizo de reversa!</p>

<pre style="background:#fce4ec; padding:0.8em; border-radius:6px;">
esta_lloviendo = True
print(not esta_lloviendo)   # Resultado: False
</pre>

<h3>📌 Consejo mágico:</h3>
<ul>
  <li><code>and</code> necesita que todo sea cierto 🧠</li>
  <li><code>or</code> se conforma con que algo sea cierto ✨</li>
  <li><code>not</code> voltea el resultado como si fuera una media 🧦</li>
</ul>

<hr style="border: none; border-top: 1px solid #80cbc4;" />

<p><em style="color:#00695c;"><strong>⚠️ Spoiler de hechicero:</strong> Estos operadores lógicos son súper importantes cuando empecemos a trabajar con <code>if</code> para tomar decisiones. Así que léelos como si fueran pociones que combinan resultados mágicos 🧪🐇</em></p>

```python
# 🧪 PRUEBA MÁGICA: Creando variables de diferentes tipos y mostrándolas con print()
# No te preocupes por modificarlas todavía, solo vamos a guardarlas y ver qué contienen

# Texto (string)
hechizo = "Abracadabra"

# Número entero (int)
edad_conejo = 5

# Número con decimales (float)
velocidad_salto = 3.5

# Valor verdadero o falso (bool)
tiene_sombrero = True

# Lista (puede guardar varios objetos)
mochila = ["varita", "zanahoria", "sombrero"]

# Tupla (igual que lista, pero no se puede modificar)
hechizos_protegidos = ("Luz", "Curación", "Invisible")

# Diccionario (clave:valor, como una mini ficha mágica)
conejo = {
    "nombre": "Saltitos",
    "edad": 3,
    "especialidad": "teletransportación"
}

# Imprimir cada uno con emojis mágicos ✨
print("📜 Hechizo:", hechizo)
print("🐰 Edad del conejo:", edad_conejo)
print("💨 Velocidad de salto:", velocidad_salto)
print("🎩 ¿Tiene sombrero?:", tiene_sombrero)
print("🎒 Mochila mágica:", mochila)
print("📚 Hechizos protegidos:", hechizos_protegidos)
print("🔖 Datos del conejo:", conejo)
```

```python
# ⚠️ Esto parecerá avanzado, pero solo es cuestión de recordarlo, o tener guardar este ejemplo para cuando lo necesites.
# Vamos a explorar cómo extraer elementos específicos y partes de una lista 🧙‍♂️
# Conjura (ejecuta) el siguiente código para ver cómo funciona:

mochila = ["varita", "zanahoria", "sombrero", "mapa", "brújula", "pañuelo mágico"]

# 1️⃣ Primer elemento (índice 0)
print("🪄 Primer objeto:", mochila[0])  # varita

# 2️⃣ Último elemento (índice -1)
print("🔚 Último objeto:", mochila[-1])  # pañuelo mágico

# 3️⃣ Segundo objeto
print("🥕 Segundo objeto:", mochila[1])  # zanahoria

# 4️⃣ Penúltimo objeto
print("⏪ Penúltimo objeto:", mochila[-2])  # brújula

# 5️⃣ Todos los objetos del segundo al cuarto (índices 1 al 3, el último no se incluye)
print("🎒 De la zanahoria al mapa:", mochila[1:4])  # ['zanahoria', 'sombrero', 'mapa']

# 6️⃣ Los tres primeros objetos
print("✨ Primeros tres objetos:", mochila[:3])  # ['varita', 'zanahoria', 'sombrero']

# 7️⃣ Los tres últimos objetos
print("🔮 Últimos tres objetos:", mochila[-3:])  # ['mapa', 'brújula', 'pañuelo mágico']

# 8️⃣ Toda la lista, pero saltando de dos en dos
print("🐾 Objetos con paso de 2:", mochila[::2])  # ['varita', 'sombrero', 'brújula']

# 9️⃣ Lista invertida (¡magia negra!)
print("🔁 Mochila al revés:", mochila[::-1])  # ['pañuelo mágico', ..., 'varita']
```

```python
# 🔧 Modificando listas mágicas
# ¡Recuerda! Las listas son como mochilas con cierre: puedes abrirlas y cambiar lo que hay dentro 🧙‍♀️🎒

mochila = ["varita", "zanahoria", "sombrero"]
print("🎒 Mochila original:", mochila)

# 1️⃣ Reemplazar un elemento (modificar)
mochila[1] = "poción de invisibilidad"
print("🔁 Mochila tras reemplazo:", mochila)

# 2️⃣ Agregar un elemento al final
mochila.append("mapa encantado")
print("➕ Agregado al final:", mochila)

# 3️⃣ Insertar un objeto en una posición específica
mochila.insert(1, "brújula mágica")
print("📍 Insertado en la posición 1:", mochila)

# 4️⃣ Eliminar un elemento por nombre
mochila.remove("sombrero")
print("❌ Quitamos el sombrero:", mochila)

# 5️⃣ Eliminar el último objeto (y guardarlo por si lo necesitamos)
objeto_eliminado = mochila.pop()
print("🧹 Eliminado el último:", objeto_eliminado)
print("🎒 Mochila final:", mochila)
```
