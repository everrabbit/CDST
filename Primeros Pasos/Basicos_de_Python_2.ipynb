<h2 style="color:#DC143C;">🧠 ¿Y qué es eso del <code>Kernel</code>? ¿Y por qué suena a robot 🤖 o del espacio 🛸?</h2>

<p><strong>Porque lo es… un poco.</strong></p>

<p>El <strong>kernel</strong> es el motorcito del notebook. Tú escribes código, él lo entiende, lo ejecuta y te responde. Es como el chef de un restaurante: tú pides “suma estos datos”, y él va, lo hace, y te entrega el platillo (el resultado).</p>

<p>Sin kernel, no pasa nada. Literal. Puedes escribir todo lo que quieras… y no obtendrás respuesta. Como gritarle a una tele apagada.</p>

<h3 style="color:#DC143C;">⚙️ ¿Dónde vive el kernel?</h3>

<ul style="margin-left: 1.5em;">
  <li><strong>Google Colab / Kaggle / otros notebooks en la nube:</strong> el kernel está en internet. Tú escribes y un servidor en el espacio cibernético ejecuta tu código.</li>
  <li><strong>Jupyter Notebook local (en tu compu):</strong> el kernel corre ahí mismo, en tu máquina. Sin Wi-Fi, pero con poder local.</li>
</ul>

<h3 style="color:#DC143C;">💥 ¿Y qué le puede pasar al kernel?</h3>

<ul style="margin-left: 1.5em;">
  <li>🔁 <strong>Se reinicia:</strong> tú lo apagaste (por error o por limpieza mental).</li>
  <li>🚫 <strong>Se cae:</strong> le pediste algo raro, como dividir entre cero o cargar un archivo de 4GB en una tostadora.</li>
  <li>😴 <strong>Se duerme:</strong> Google decidió que ya fue mucho por hoy.</li>
</ul>

<p>¿La solución mágica? <strong>Reinicia el kernel</strong>. Ya.  
Haz clic en “<em>Restart kernel</em>” o su equivalente, y todo vuelve a la normalidad.  
Es el “apaga y prende” de toda la vida. Funciona 9 de cada 10 veces.</p>

## 🧩 ¿Qué es una celda y cómo se ejecuta?

Ya tienes tu entorno listo, abriste un notebook, y… de pronto te enfrentas a un rectángulo en blanco.  
¡Esa es una **celda**!

En un notebook trabajas celda por celda, como si fueran bloques de instrucciones. Y hay de dos sabores principales:

- 💻 **Celda de código**: aquí va el Python. Lo que escribas aquí se puede ejecutar (si el kernel está activo).
- 📝 **Celda Markdown**: sirve para escribir texto bonito, títulos, listas, imágenes, tablas... como este tutorial.

---

### 🧠 ¿Y si dice “Python” pero no funciona?

A veces verás arriba de la celda algo como `Python` pero en gris o como “No Code”.  
Eso significa que el tipo de celda no está correctamente asignado. Puede que:

- Sea una celda Markdown disfrazada de código.
- No esté asociada al kernel.
- O accidentalmente se haya cambiado el lenguaje a otro.

---

✅ **¿Cómo saber que estás en una celda de código real?**

- Asegúrate de que en la esquina superior izquierda de la celda diga `Code` (no Markdown).
- En VSCode, revisa el menú desplegable que está justo arriba a la izquierda de cada celda.
- Si aparece algo raro como “Plain text” o “No Kernel”, puedes hacer clic y cambiarlo a `Python`.

> 📌 **Dato importante**: si escribes código en una celda mal configurada… no va a pasar nada.

## 🧩 ¿Con qué se construye Python?

Imagina que vas a enseñarle a alguien cómo preparar un sándwich, paso a paso.  
Así funciona Python: tú le das instrucciones, una por una.

Algunas piezas básicas son:

---

### 📦 1. Variables  
Una **variable** es como una cajita donde guardas algo.  
Le pones nombre a la caja y metes un valor dentro. ¿Pero de qué me sirve guardar cosas en cajas?
Porque luego puedes usar ese valor en tu código, sin tener que escribirlo de nuevo.

Entonces para que Python entienda que quieres guardar algo, usas el símbolo debes tener presente 3 principales cosas:
1. **Nombre de la variable**: como una etiqueta para tu caja.
2. **Símbolo de asignación**: el signo igual `=` que le dice a Python que guarde algo en esa caja.
3. **Valor**: lo que realmente quieres guardar, como un número, texto o lista (esta última la veremos más adelante).

Imaginemos que queremos guardar la el valor de pi, naaa, mejor algo más sencillo como la edad de una persona en años. No necesariamente lo debes pensar así, pero para que quede claro, lo haremos paso a paso:

- **Paso 1** Como soy un poco flojo y no quiero pensar mucho, nombraré **edad** a mi variable.
- **Paso 2** Necesito el operador igual **=**  (🤡 no hay que pensar en nada, simplemente lo agregas donde va)
- **Paso 3** Y por último, el valor que quiero guardar, en este caso, digamos que la edad es 25 años (ya quisiera tener 25 años otra vez 😅).

¿Cómo se lo digo a Python? Muy fácil, así:

```python
edad = 25
```

  <p>El símbolo <code>=</code> aquí no significa “igual” como en matemáticas.  
  👉 En Python, el <code>=</code> quiere decir: “guarda esto dentro de esta caja (variable)”.</p>
</div>

<h2 style="color:#ff7043;">✏️ ¿Cómo se nombra una variable correctamente?</h2>

<p>Imagina que cada variable es una cajita mágica con un nombre. Si no la nombras bien, Python no sabrá qué hacer… y lanzará un grito de error 😱.</p>

<h3>📌 Reglas básicas para nombrar variables:</h3>
<ul>
  <li>✅ Solo usa <strong>letras</strong>, <strong>números</strong> y <code>_</code> (guión bajo)</li>
  <li>❌ <strong>No puede comenzar con un número</strong> (¡prohibido por el Consejo Conejil!)</li>
  <li>❌ <strong>No uses espacios</strong> (usa <code>_</code> para separar palabras)</li>
  <li>❌ <strong>No uses tildes, ñ, signos raros ni emojis</strong> en el nombre</li>
  <li>❌ <strong>No uses palabras reservadas</strong> (como <code>print</code>, <code>if</code>, <code>True</code>)</li>
</ul>

<h3>✅ Ejemplos válidos:</h3>

<pre style="background:#ffe0b2; padding:0.8em; border-radius:6px;">
nombre_conejo = "Saltitos"
edad = 3
pocion_magica_1 = "Invisibilidad"
tiene_sombrero = True
</pre>

<h3>❌ Ejemplos no válidos:</h3>

<pre style="background:#ef9a9a; padding:0.8em; border-radius:6px;">
123vida = "Error"         # ❌ Empieza con número
nombre conejo = "Error"   # ❌ Tiene espacio
edad? = 3                 # ❌ Signo no permitido
print = "Error"           # ❌ Palabra reservada
</pre>

<h3>🤔 ¿Mayúsculas, minúsculas, guiones?</h3>
<ul>
  <li><strong>Python distingue mayúsculas de minúsculas</strong> → <code>magia</code> no es lo mismo que <code>Magia</code></li>
  <li>Por convención, usamos <strong>minúsculas con guión bajo</strong>: <code>nivel_magico</code></li>
</ul>

<h2 style="color:#ff7043;">🥕 Variables enteras (int)</h2>

<p>A veces, un conejo solo necesita contar cosas completas: <strong>zanahorias, saltos, sombreros, hechizos lanzados</strong>. Cosas que no tienen mitades. Para eso usamos las variables de tipo <code>int</code>, que viene de <strong>integer</strong>, o sea, número entero.</p>

<h3>🧮 ¿Cómo se escribe un número entero?</h3>
<p>Sin comillas, sin punto, sin adornos:</p>

<pre style="background:#ffe0b2; padding:0.8em; border-radius:6px;">
zanahorias = 7
saltos_realizados = 12
hechizos_pendientes = 0
</pre>

<h3>📏 ¿Y los números muy grandes?</h3>
<p>¡También se puede! Y puedes separarlos con guiones bajos para que se entiendan mejor:</p>

<pre style="background:#ffe0b2; padding:0.8em; border-radius:6px;">
poblacion_conejos = 1_000_000_000   # Mucho más fácil de leer que 1000000000
</pre>

<p>Python ignora los guiones bajos, pero tú los verás clarito.</p>

<h3>❌ ¿Y si lo escribo como texto?</h3>

<pre style="background:#ffcdd2; padding:0.8em; border-radius:6px;">
edad = "7"   # ❌ Esto es texto, no número
</pre>

<p>Si el número va entre comillas, <strong>ya no es un número</strong>: es un texto. Y no podrás hacer operaciones mágicas con él (como sumarlo o dividirlo).</p>

<h3>📦 ¿Qué se puede guardar como entero?</h3>
<ul>
  <li>Cantidades completas: <code>3</code>, <code>0</code>, <code>99</code></li>
  <li>Números negativos también sirven: <code>-5</code></li>
  <li>Números grandes con guiones bajos para que se entiendan mejor</li>
</ul>

<h3>🔍 ¿Cómo saber si algo es un int?</h3>
<p>Si no tiene punto (<code>.</code>) y no está entre comillas, seguramente es un entero.</p>

<pre style="background:#ffe0b2; padding:0.8em; border-radius:6px;">
zanahorias_comidas = 4       # ✅ int
zanahorias_comidas = 4.0     # ✅ pero esto es float
zanahorias_comidas = "4"     # ❌ texto
</pre>

<h3>🧠 ¿Cuándo usar int?</h3>
<p>Cuando cuentes objetos, personas, intentos o repeticiones. Todo lo que no tenga mitades mágicas ni fracciones.</p>

<h2 style="color:#00838f;">💧 Variables con decimales (float)</h2>

<p>Hay ocasiones en que un conejo necesita más precisión mágica. Por ejemplo, calcular el salto exacto desde la madriguera, o medir la temperatura de una poción burbujeante.</p>

<p>Para eso existen las variables de tipo <strong>float</strong> (abreviación de <em>floating point</em>, en inglés), que permiten guardar <strong>números con punto decimal</strong>.</p>

<h3>📏 ¿Cómo se escribe un número con decimales?</h3>
<p>Solo agrégale el punto (<code>.</code>):</p>

<pre style="background:#fce4ec; padding:0.8em; border-radius:6px;">
salto_metros = 2.5
temperatura_pocion = 37.8
</pre>

<h3>❌ ¿Y si uso una coma en lugar del punto?</h3>

<pre style="background:#ffcdd2; padding:0.8em; border-radius:6px;">
salto = 1,75   # ❌ ¡Ups! Python cree que estás guardando dos cosas separadas
</pre>

<p>Recuerda: <strong>los decimales usan punto, no coma</strong>. Aunque en el mundo muggle usamos coma, Python prefiere el estilo anglosajón. Así que será <code>2.5</code>, no <code>2,5</code>.</p>

<h3>🔍 ¿Cómo sé que una variable es tipo float?</h3>
<ul>
  <li>Siempre contiene un <strong>punto decimal</strong>, incluso si es <code>3.0</code></li>
  <li>Puedes hacer operaciones matemáticas más precisas con ellas</li>
</ul>

<h3>🧠 ¿Cuándo usar float y no int?</h3>
<p>Cuando necesites fracciones, promedios, porcentajes, distancias, temperaturas o cualquier cosa que no sea un número entero.</p>

<pre style="background:#fce4ec; padding:0.8em; border-radius:6px;">
promedio_notas = 8.75
peso_zanahoria = 0.25
</pre>

<hr style="border: none; border-top: 1px solid #80deea;" />

<p><em style="color:#00796b;"><strong>📌 Consejo mágico:</strong> Si necesitas más precisión que “solo un número redondo”, elige float. Y recuerda usar el punto 🐰.</em></p>

<h2 style="color:#00838f;">📜 Variables de tipo texto</h2>

<p>En el mundo de la programación, cuando se quiere guardar palabras, frases o incluso hechizos, las pone dentro de una cajita especial llamada <strong>string</strong> (así en inglés).</p>

<p>¿Cómo sabe el sistema que lo que guardaste son <strong>palabras</strong> y no números? Fácil: <strong>si está entre comillas, es texto</strong>.</p>

<h3>✨ ¿Qué tipo de comillas puedo usar?</h3>
<ul>
  <li><code>"comillas dobles"</code></li>
  <li><code>'comillas simples'</code></li>
  <li><code>"""tres comillas para varias líneas"""</code></li>
</ul>

<h3>🪄 Ejemplos de hechizos válidos:</h3>

<pre style="background:#fce4ec; padding:0.8em; border-radius:6px;">
hechizo = "Abracadabra"
mensaje = 'El mago conejo llegó'
pocion = """Esta es una receta mágica
que ocupa varias líneas"""
</pre>

<h3>❌ ¿Y si no uso comillas?</h3>

<pre style="background:#ffcdd2; padding:0.8em; border-radius:6px;">
hechizo = Abracadabra   # ❌ ERROR: el sistema no entiende qué es eso
</pre>

<p><strong>Siempre</strong> que quieras guardar texto (como frases, palabras, encantamientos o nombres de conejos), <strong>usa comillas</strong>. Si no lo haces, el sistema se confunde y lanza chispas.</p>

<h3>📦 ¿Qué se puede guardar como texto?</h3>
<ul>
  <li>Palabras: <code>"zanahoria"</code></li>
  <li>Frases: <code>"Salto mágico nivel 2"</code></li>
  <li>Números que quieres tratar como texto: <code>"1234"</code></li>
  <li>Incluso emojis y símbolos: <code>"🧙‍♂️🐰✨"</code></li>
</ul>

<h3>🧠 Cajitas mágicas multilínea</h3>
<p>Si necesitas escribir un hechizo largo, que ocupa más de una línea, usa <code>"""triple comilla</code>:</p>

<pre style="background:#fce4ec; padding:0.8em; border-radius:6px;">
manual = """Paso 1: Agita la varita
Paso 2: Mira al conejo

</pre>

<p>Esto es muy útil para dejar mensajes, anotar explicaciones o escribir hechizos con pasos 🪄.</p>

<h2 style="color:#f57c00;">🔮 Variable tipo booleana: <code>True</code> o <code>False</code> (<code>bool</code>)</h2>

<p>Hay preguntas que no necesitan números, ni textos largos... solo una respuesta simple:</p>

<ul>
  <li>¿El hechizo funcionó?</li>
  <li>¿El conejo tiene capa?</li>
  <li>¿El caldero está vacío?</li>
</ul>

<p>En esos casos, usamos una variable <strong>booleana</strong> (sí, se llama así), del tipo <code>bool</code>. Solo acepta dos valores posibles:</p>

<ul>
  <li><code>True</code> — verdadero (sí, afirmativo, magia activada)</li>
  <li><code>False</code> — falso (no, negativo, hechizo fallido)</li>
</ul>

<h3>🎯 Ejemplos conejiles:</h3>

<pre style="background:#ffe0b2; padding:0.8em; border-radius:6px;">
hechizo_exitoso = True
tiene_varita = False
conejera_llena = True
</pre>

<h3>⚠️ ¡Importante!</h3>
<ul>
  <li><code>True</code> y <code>False</code> van con mayúscula inicial</li>
  <li>No llevan comillas</li>
</ul>

<pre style="background:#ffcdd2; padding:0.8em; border-radius:6px;">
hechizo_exitoso = "True"   # ❌ Esto es texto, no magia real
</pre>

<h3>🧠 ¿Cómo se usan?</h3>
<p>Las variables booleanas se usan para tomar decisiones en tu código. Como decir: <em>“si esto es verdadero, haz algo”</em>.</p>

<p>Por ahora lo haremos simple, con una comparación explícita:</p>

<pre style="background:#fce4ec; padding:0.8em; border-radius:6px;">
if hechizo_exitoso == True:
    print("🎉 ¡El conejo desapareció!")
</pre>

<h2 style="color:#f06292;">🧺 Cajitas mágicas que guardan varias cosas (<code>list</code>)</h2>

<p>Imagina que el conejo aprendiz tiene una bolsa encantada donde guarda varias cosas: una varita, una zanahoria, un sombrero... Todo en el mismo lugar, en orden.</p>

<p>Esa bolsita mágica es una <strong>lista</strong> en Python.</p>

<h3>📦 ¿Cómo se escribe una lista?</h3>
<p>Usa <strong>corchetes</strong> <code>[ ]</code> y separa los elementos con comas.</p>

<pre style="background:#f8bbd0; padding:0.8em; border-radius:6px;">
bolsa = ["zanahoria", "sombrero", "varita"]
</pre>

<h3>🧠 Cosas importantes que debes saber:</h3>
<ul>
  <li>Las listas pueden guardar texto, números, o mezclas mágicas</li>
  <li>Siempre van entre <strong>corchetes</strong> <code>[ ]</code></li>
  <li>Los elementos van separados por <strong>comas</strong></li>
  <li>Puede estar vacía: <code>[]</code></li>
</ul>

<h3>🎯 Ejemplos:</h3>

<pre style="background:#f8bbd0; padding:0.8em; border-radius:6px;">
hechizos = ["Invisibilidad", "Teletransportación", "Luz"]
numeros = [1, 2, 3, 4]
mezcla = [True, "mapa", 8.5]
</pre>

<h3>❌ Algunos errores comunes:</h3>

<h3>😵‍💫 ¿Y si no uso comillas?</h3>
<p>Recuerda que si quieres guardar palabras o frases directamente, debes escribirlas <strong>entre comillas</strong>:</p>

<pre style="background:#ffcdd2; padding:0.8em; border-radius:6px;">
hechizos = [Invisibilidad, Teletransportación]   # ❌ ERROR: Python busca variables con esos nombres
</pre>

<p>Este error pasa porque Python cree que <code>Invisibilidad</code> y <code>Teletransportación</code> son <em>variables</em> que ya existen. Si no las has creado antes, lanzará chispas mágicas (o mejor dicho, un <code>NameError</code>).</p>

<p>✅ Lo correcto si quieres guardar texto directamente:</p>

<pre style="background:#f8bbd0; padding:0.8em; border-radius:6px;">
hechizos = ["Invisibilidad", "Teletransportación"]
</pre>

<h3>🌟 ¿Y si sí quiero guardar variables?</h3>
<p>También puedes hacerlo. Si ya definiste variables antes, puedes meterlas en una lista así:</p>

<pre style="background:#f8bbd0; padding:0.8em; border-radius:6px;">
pocion_1 = "Curación"
pocion_2 = "Invisibilidad"
pocion_3 = "Fuerza"

inventario = [pocion_1, pocion_2, pocion_3]   # ✅ Lista con variables
</pre>

<p>En resumen: <strong>si son textos, pon comillas. Si son variables que ya existen, no las necesitas.</strong> Pero si escribes palabras sin comillas y no existen… ¡se arma el hechicidio!</p>


<h3>🔍 ¿Qué puedo hacer con una lista?</h3>
<ul>
  <li>Guardar muchos valores en una sola variable</li>
  <li>Agregar, quitar o cambiar cosas mágicamente (más adelante lo verás)</li>
  <li>Revisar qué hay en la lista... y sacar cosas una por una</li>
</ul>

<p>Ejemplo de cómo ver lo que contiene:</p>

<pre style="background:#f8bbd0; padding:0.8em; border-radius:6px;">
print(hechizos)  
# Resultado: ['Invisibilidad', 'Teletransportación', 'Luz']
</pre>

<h3>🔧 ¿Las listas se pueden modificar?</h3>

<p>¡Sí! Las listas en Python son como mochilas mágicas con cierre: puedes abrirlas, meter cosas nuevas, sacar lo que ya no sirve o cambiar algo que está dentro.</p>

<p>Por ahora no necesitas aprender cómo se hace, pero <strong>es importante saber que se puede</strong>.</p>

<blockquote style="background:#ffe0b2; padding:0.7em; border-left: 4px solid #f57c00; border-radius: 6px;">
Las listas <strong>sí son modificables</strong> (o como dice la magia técnica: son <em>mutables</em>).
</blockquote>

<p>Más adelante aprenderás a:</p>
<ul>
  <li>Agregar un hechizo nuevo a la lista</li>
  <li>Quitar uno que salió defectuoso</li>
  <li>Reemplazar ingredientes mágicos</li>
</ul>

<p>Pero por ahora, quédate con esto:</p>

<pre style="background:#f8bbd0; padding:0.8em; border-radius:6px;">
inventario = ["zanahoria", "sombrero", "varita"]
# Puedes modificar esta lista en cualquier momento
</pre>

<h2 style="color:#9575cd;">🔐 Listas selladas con hechizo (<code>tuple</code>)</h2>

<p>Una <strong>tupla</strong> es como una lista, pero cerrada con magia. Una vez que la creas, <strong>no puedes cambiarla</strong>. Ni sacar ni meter nada. Ni modificar lo que hay adentro. Es como si le pusieras un sello mágico que dice: “esto ya no se toca” 🪄</p>

<h3>📦 ¿Cómo se escribe una tupla?</h3>
<p>Usa <strong>paréntesis</strong> <code>( )</code> y separa los elementos con comas, como en una lista.</p>

<pre style="background:#d1c4e9; padding:0.8em; border-radius:6px;">
hechizos_fijos = ("Invisibilidad", "Luz", "Curación")
</pre>

<h3>🧠 Cosas importantes:</h3>
<ul>
  <li>Van entre <strong>paréntesis</strong> <code>( )</code></li>
  <li>Sus elementos están separados por comas</li>
  <li>Una vez creada, <strong>no se puede cambiar</strong></li>
</ul>

<h3>📌 ¿Para qué sirve una tupla?</h3>

<h3>🤖 ¿Y si intento cambiar algo?</h3>

<pre style="background:#ffcdd2; padding:0.8em; border-radius:6px;">
elementos = ("agua", "fuego", "aire")

</pre>

<p>Python lanzará un hechizo de error llamado <code>TypeError</code>, y te dirá que estás tratando de hacer algo prohibido 🧙‍♂️.</p>

<h2 style="color:#4db6ac;">📖 Diccionarios mágicos (<code>dict</code>)</h2>

<p>Un <strong>diccionario</strong> en Python es como un libro de hechizos o un archivo secreto de conejos magos. Guarda información organizada por <strong>palabras clave</strong> y sus <strong>valores</strong>.</p>

<p>Es como decir:</p>

<blockquote style="background:#b2dfdb; padding:0.7em; border-left: 4px solid #4db6ac; border-radius: 6px;">
"El conejo se llama <strong>Saltitos</strong>, tiene <strong>3 años</strong>, y su especialidad es <strong>teletransportación</strong>"
</blockquote>

<h3>🔧 ¿Cómo se escribe un diccionario?</h3>
<p>Usa <strong>llaves</strong> <code>{ }</code>, y dentro, cada dato se escribe como: <code>"clave": valor</code></p>

<pre style="background:#b2dfdb; padding:0.8em; border-radius:6px;">
conejo = {
  "nombre": "Saltitos",
  "edad": 3,
  "especialidad": "teletransportación"
}
</pre>

<h3>🧙‍♀️ ¿Y esos dos puntos <code>:</code> qué hacen?</h3>

<p>En un diccionario, los <strong>dos puntos</strong> <code>:</code> son la varita que conecta una <strong>clave</strong> con su <strong>valor</strong>.</p>

<blockquote style="background:#b2dfdb; padding:0.7em; border-left: 4px solid #4db6ac; border-radius: 6px;">
Imagina que estás diciendo: <br>
<strong>"La clave 'nombre' tiene como valor 'Saltitos'"</strong>
</blockquote>

<p>Así se ve en código:</p>

<pre style="background:#b2dfdb; padding:0.8em; border-radius:6px;">
conejo = {
  "nombre": "Saltitos",      # clave: valor
  "edad": 3,
  "especialidad": "teletransportación"
}
</pre>

<p>Sin los <code>:</code>, Python no sabría qué información corresponde a qué palabra clave… y se haría bolas mágicas 🌀.</p>


<h3>🧠 Cosas importantes que debes saber:</h3>
<ul>
  <li>Las <strong>claves</strong> suelen ir entre comillas (texto)</li>
  <li>Los <strong>valores</strong> pueden ser textos, números, booleanos… incluso listas o más diccionarios</li>
  <li>Cada par clave-valor se separa por comas</li>
</ul>

<h3>🔍 ¿Para qué sirve un diccionario?</h3>
<ul>
  <li>Para guardar información etiquetada</li>
  <li>Cuando necesitas acceder a datos por su nombre, no por posición</li>
  <li>Ideal para representar objetos mágicos, personajes, configuraciones…</li>
</ul>

<h3>🎯 Ejemplo útil:</h3>

<pre style="background:#b2dfdb; padding:0.8em; border-radius:6px;">
hechizo = {
  "nombre": "Luz",
  "nivel": 1,
  "activo": True,
  "ingredientes": ["vela", "esencia de luna"]
}
</pre>

<h3>🚨 Algunos errores comunes:</h3>

<pre style="background:#ffcdd2; padding:0.8em; border-radius:6px;">
# ❌ Faltan comillas en la clave
mago = {nombre: "Bunelius"}   # Python pensará que 'nombre' es una variable
</pre>

<p>✅ Correcto:</p>

<pre style="background:#b2dfdb; padding:0.8em; border-radius:6px;">
mago = {"nombre": "Bunelius"}
</pre>

<h3>🧠 Bonus mágico: ¿cómo accedo a un dato?</h3>

<p>Como si buscaras una página específica del libro:</p>

<pre style="background:#fce4ec; padding:0.8em; border-radius:6px;">
print(conejo["nombre"])  # Resultado: Saltitos
</pre>

<p>No puedes usarlas como nombres de tus variables. Son <strong>palabras reservadas</strong> y ya están ocupadas 🧙‍♀️🔒</p>

<pre style="background:#b2dfdb; padding:0.8em; border-radius:6px;">

# **Métodos y funciones**

<div style="border-left: 6px solid #26c6da; padding: 1.5em; background: #e0f7fa; border-radius: 12px; font-family: 'Segoe UI', sans-serif; color: #004d40; line-height: 1.6;">

<h2 style="color:#00838f;">🔮 Métodos: acciones que viven dentro de tus variables</h2>

<p>Hay ciertas cajitas mágicas (como listas, textos o números) que tienen <strong>hechizos secretos incorporados</strong>. A esos hechizos les llamamos <strong>métodos</strong>.</p>

<p>Un método es como decirle a una variable: “¡haz algo contigo misma!”</p>

<p>Y se ve así:</p>

<pre style="background:#b2ebf2; padding:0.8em; border-radius:6px;">
nombre_variable.metodo()
</pre>

<h3>✨ Ejemplo real:</h3>

<pre style="background:#e0f2f1; padding:0.8em; border-radius:6px;">
mensaje = "abracadabra"
print(mensaje.upper())    # Resultado: "ABRACADABRA"
</pre>

<p>Acá, <code>.upper()</code> es un método que convierte el texto a mayúsculas. Y como está pegado con un <code>.</code> a la variable <code>mensaje</code>, ¡sabemos que es un método!</p>

<h3>📦 ¿Y los paréntesis?</h3>

<p>Todos los métodos <strong>llevan paréntesis</strong> porque pueden necesitar instrucciones adicionales (llamadas <em>parámetros</em>).</p>

<ul>
  <li><strong>Algunos métodos no necesitan nada:</strong> <code>mochila.clear()</code></li>
  <li><strong>Otros sí:</strong> <code>texto.replace("a", "e")</code></li>
</ul>

<h3>🤔 ¿Y cómo sé que es un método y no una función?</h3>

  <thead style="background:#b2dfdb;">

    <tr style="background:#f1f8e9;">

<h2 style="color:#f4511e;">🧪 ¿Qué es una función?</h2>

<p>Una función es como un <strong>hechizo reutilizable</strong>: le das una orden a Python para que haga algo... <em>cada vez que invoques ese hechizo</em>.</p>

<p>Las funciones <strong>siempre se escriben con paréntesis</strong> (<code>()</code>) y a veces llevan ingredientes dentro (también conocidos como <em>parámetros</em>).</p>

<h3>🎩 Hay dos tipos de funciones:</h3>

<ul>
  <li><strong>Funciones de Python</strong> (ya vienen listas)</li>
  <li><strong>Funciones mágicas propias</strong> (las puedes crear tú más adelante)</li>
</ul>

<h3>✨ Ejemplos de funciones que ya existen en Python:</h3>

<pre style="background:#ffe0b2; padding:0.8em; border-radius:6px;">
print("Hola mundo")        # Muestra algo en pantalla
len("Abracadabra")         # Cuenta cuántas letras hay
type(3.14)                 # Dice de qué tipo es un dato
</pre>

<h3>🔮 ¿Y cómo se ve una función inventada?</h3>

<pre style="background:#ffccbc; padding:0.8em; border-radius:6px;">
def saludar():
    print("¡Hola aprendiz de magia!")

saludar()
</pre>

<p>En este caso, tú mismo creaste una función llamada <code>saludar()</code>. Cuando la invocas, ¡hace su hechizo!</p>

<h3>📦 ¿Qué hay dentro de los paréntesis?</h3>

<p>Las funciones pueden aceptar uno o más ingredientes (parámetros) para trabajar. Pero también pueden estar vacías, como un hechizo simple:</p>

<pre style="background:#fff3e0; padding:0.8em; border-radius:6px;">
print("Conejo feliz")      # lleva texto como ingrediente
print()                    # no lleva nada, pero igual funciona
</pre>

<h3>🧠 ¿Cómo reconozco que es una función?</h3>
<ul>
  <li>Tiene nombre (como <code>print</code> o <code>len</code>)</li>
  <li>Sigue un par de paréntesis <code>()</code></li>
  <li>¡No lleva punto antes! (eso es de los métodos)</li>
</ul>

<hr style="border: none; border-top: 1px solid #ffab91;" />

<p><em style="color:#bf360c;"><strong>📌 Consejo mágico:</strong> Si ves algo como <code>algo()</code> sin punto y con paréntesis, probablemente es una función. Puede ser una de las funciones que Python ya trae en su grimorio... o una que tú creaste con tu varita ✨</em></p>

</div>

<h2 style="color:#00897b;">🥄 Parámetros: lo que va dentro de los paréntesis</h2>

<p>Cuando ves algo como <code>algo()</code> en Python, esos paréntesis pueden ir vacíos… o llevar <strong>ingredientes</strong>, como si fuera una receta mágica.</p>

<p>A esos ingredientes les llamamos <strong>parámetros</strong>. Puedes poner uno, dos, tres o más. ¡Solo recuerda separarlos con comas!</p>

<h3>🎩 Ejemplo con función (Python ya la trae):</h3>

<pre style="background:#dcedc8; padding:0.8em; border-radius:6px;">
print("Hola", "aprendiz", 2025)
</pre>

<p>Esta función está mostrando tres cosas diferentes a la vez. Cada una va separada por una coma, ¡y no tienen que ser del mismo tipo!</p>

<h3>📦 Ejemplo con método (propio de una variable):</h3>

<pre style="background:#b2ebf2; padding:0.8em; border-radius:6px;">
frase = "El conejo saltó"
nueva_frase = frase.replace("conejo", "hechicero")
print(nueva_frase)   # Resultado: "El hechicero saltó"
</pre>

<p>El método <code>.replace()</code> lleva dos parámetros: lo que quieres reemplazar y con qué lo quieres cambiar. Cada uno es un ingrediente del hechizo.</p>

<h3>🍡 Pueden ser textos, números, listas, booleanos... lo que sea:</h3>

<pre style="background:#f0f4c3; padding:0.8em; border-radius:6px;">
ingredientes = ["zanahoria", "poción", "varita"]
print("🧪 Mochila mágica contiene:", ingredientes, True)
</pre>

<h3>⚠️ ¿Y si le doy un parámetro raro?</h3>

<p>Hay funciones o métodos que <strong>esperan un tipo de dato específico</strong>. Por ejemplo, puede que solo funcionen con texto, o con números, o con listas. Si les das algo diferente, Python se confunde… y lanza un grito (también llamado error 🐍💥).</p>

<pre style="background:#ffcdd2; padding:0.8em; border-radius:6px;">
numero = 42
print(numero.replace("2", "9"))  # ❌ ERROR: los números no tienen .replace()
</pre>

<p>En este caso, <code>.replace()</code> es un método exclusivo de los textos (strings). El número <code>42</code> no sabe qué hacer con eso, así que Python se molesta y te lo dice.</p>

<h3>🧪 En funciones propias también se puede indicar el tipo</h3>

<p>Cuando crees tus propias funciones (más adelante), también podrás <strong>especificar qué tipo de datos esperas recibir</strong>. Y si no te dan lo correcto, puedes hacer que la función proteste, corrija o falle.</p>

<p><em style="color:#00695c;"><strong>📌 Consejo mágico:</strong> Cada hechizo necesita los ingredientes correctos. Si la receta dice “zanahoria” y tú pones una varita... algo va a explotar 🧨</em></p>

<p><em style="color:#00695c;"><strong>📌 Consejo mágico:</strong> Si ves una lista de cosas separadas por comas dentro de paréntesis, esos son parámetros. Están ahí para darle instrucciones más precisas a tu función o método 🧪🐰</em></p>

</div>


````python
# 🧪 PRUEBA MÁGICA: Creando variables de diferentes tipos y mostrándolas con print()
# No te preocupes por modificarlas todavía, solo vamos a guardarlas y ver qué contienen

# Texto (string)
hechizo = "Abracadabra"

# Número entero (int)
edad_conejo = 5

# Número con decimales (float)
velocidad_salto = 3.5

# Valor verdadero o falso (bool)
tiene_sombrero = True

# Lista (puede guardar varios objetos)
mochila = ["varita", "zanahoria", "sombrero"]

# Tupla (igual que lista, pero no se puede modificar)
hechizos_protegidos = ("Luz", "Curación", "Invisible")

# Diccionario (clave:valor, como una mini ficha mágica)
conejo = {
    "nombre": "Saltitos",
    "edad": 3,
    "especialidad": "teletransportación"
}

# Imprimir cada uno con emojis mágicos ✨
print("📜 Hechizo:", hechizo)
print("🐰 Edad del conejo:", edad_conejo)
print("💨 Velocidad de salto:", velocidad_salto)
print("🎩 ¿Tiene sombrero?:", tiene_sombrero)
print("🎒 Mochila mágica:", mochila)
print("📚 Hechizos protegidos:", hechizos_protegidos)
print("🔖 Datos del conejo:", conejo)
````

<h2 style="color:#f06292;">🔍 Cómo acceder a elementos en una <code>list</code></h2>

<p>Imagina que tu lista mágica es como una fila de casilleros con pociones. Cada casillero tiene un número (invisible a simple vista) que indica su posición. A ese número se le llama <strong>índice</strong>.</p>

<p>Y aquí viene el detalle curioso: <strong>Python empieza a contar desde cero</strong> 🧙‍♂️. Así que el primer elemento está en la posición 0, el segundo en la 1, y así sucesivamente.</p>

<h3>📦 Ejemplo:</h3>

<pre style="background:#f8bbd0; padding:0.8em; border-radius:6px;">
mochila = ["varita", "zanahoria", "sombrero", "mapa"]
</pre>

<h3>🪄 ¿Cómo se accede a un elemento?</h3>

<p>Usa el nombre de la lista seguido del número del índice entre corchetes <code>[ ]</code>:</p>

<pre style="background:#f8bbd0; padding:0.8em; border-radius:6px;">
print(mochila[0])   # Resultado: varita
print(mochila[2])   # Resultado: sombrero
</pre>

<h3>😬 ¿Qué pasa si me paso del límite?</h3>

<pre style="background:#ffcdd2; padding:0.8em; border-radius:6px;">
print(mochila[10])  # ❌ ERROR: no existe ese índice
</pre>

<p>Python te avisará con un <code>IndexError</code>, que básicamente significa: “No hay nada en ese casillero” 🧳🚫</p>

<hr style="border: none; border-top: 1px solid #f48fb1;" />

<p><em style="color:#ad1457;"><strong>📌 Consejo mágico:</strong> 
Cuando veas algo como <code>lista[1]</code>, recuerda que está pidiendo el segundo elemento. Porque en este mundo encantado… ¡empezamos desde el 0!</em></p>

</div>


````python
# ⚠️ Esto parecerá avanzado, pero solo es cuestión de recordarlo, o tener guardar este ejemplo para cuando lo necesites.
# Vamos a explorar cómo extraer elementos específicos y partes de una lista 🧙‍♂️
# Conjura (ejecuta) el siguiente código para ver cómo funciona:

mochila = ["varita", "zanahoria", "sombrero", "mapa", "brújula", "pañuelo mágico"]

# 1️⃣ Primer elemento (índice 0)
print("🪄 Primer objeto:", mochila[0])  # varita

# 2️⃣ Último elemento (índice -1)
print("🔚 Último objeto:", mochila[-1])  # pañuelo mágico

# 3️⃣ Segundo objeto
print("🥕 Segundo objeto:", mochila[1])  # zanahoria

# 4️⃣ Penúltimo objeto
print("⏪ Penúltimo objeto:", mochila[-2])  # brújula

# 5️⃣ Todos los objetos del segundo al cuarto (índices 1 al 3, el último no se incluye)
print("🎒 De la zanahoria al mapa:", mochila[1:4])  # ['zanahoria', 'sombrero', 'mapa']

# 6️⃣ Los tres primeros objetos
print("✨ Primeros tres objetos:", mochila[:3])  # ['varita', 'zanahoria', 'sombrero']

# 7️⃣ Los tres últimos objetos
print("🔮 Últimos tres objetos:", mochila[-3:])  # ['mapa', 'brújula', 'pañuelo mágico']

# 8️⃣ Toda la lista, pero saltando de dos en dos
print("🐾 Objetos con paso de 2:", mochila[::2])  # ['varita', 'sombrero', 'brújula']

# 9️⃣ Lista invertida (¡magia negra!)
print("🔁 Mochila al revés:", mochila[::-1])  # ['pañuelo mágico', ..., 'varita']
````

````python
# 🔧 Modificando listas mágicas
# ¡Recuerda! Las listas son como mochilas con cierre: puedes abrirlas y cambiar lo que hay dentro 🧙‍♀️🎒

mochila = ["varita", "zanahoria", "sombrero"]
print("🎒 Mochila original:", mochila)

# 1️⃣ Reemplazar un elemento (modificar)
mochila[1] = "poción de invisibilidad"
print("🔁 Mochila tras reemplazo:", mochila)

# 2️⃣ Agregar un elemento al final
mochila.append("mapa encantado")
print("➕ Agregado al final:", mochila)

# 3️⃣ Insertar un objeto en una posición específica
mochila.insert(1, "brújula mágica")
print("📍 Insertado en la posición 1:", mochila)

# 4️⃣ Eliminar un elemento por nombre
mochila.remove("sombrero")
print("❌ Quitamos el sombrero:", mochila)

# 5️⃣ Eliminar el último objeto (y guardarlo por si lo necesitamos)
objeto_eliminado = mochila.pop()
print("🧹 Eliminado el último:", objeto_eliminado)
print("🎒 Mochila final:", mochila)
````
