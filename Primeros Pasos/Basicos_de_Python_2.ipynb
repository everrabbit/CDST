## ➕ Suma - Juntando conejos

La suma nos ayuda a juntar cosas. Por ejemplo, si tenemos 3 conejos blancos y llegan 2 conejos negros, ¿cuántos conejos tenemos en total?

**Símbolo:** +

**Ejemplo:** 3 + 2 = 5 (ahora tenemos 5 conejos en total)

</div>

## Python permite sumar números de varias formas:

### - Una suma directa de dos números

```python
# Ejecuta el código para que veas el resultado
2 + 5
```

### - Si existen variables con valores numéricos, también se pueden sumar, estos valores pueden ser enteros o decimales. Crearemos dos variable y les asignaremos valores numéricos

```python
conejos_blancos = 10
conejos_negros = 5
```

### Ahora sumaremos los valores

```python
conejos_blancos + conejos_negros
```

### - Valores de una variable + un valor numerico directo

```python
conejos_negros + 30
```

### - Esto no limita a que solo se deban sumar dos valores al mismo tiempo, no buscaremos una regla de cuántos pero considera que puedes sumar muchos

```python
conejos_negros + conejos_blancos + 30 + 100 + 500 + 20000
```

<div style="border: 3px solid #FF9800; padding: 15px; border-radius: 10px; background-color: #FFF3E0;">

# ➖ Resta - Cuando se van algunos conejos

La resta nos ayuda a quitar cosas. Por ejemplo, si tenemos 8 conejos y se escapan 3, ¿cuántos conejos nos quedan?

**Símbolo:** -

**Ejemplo:** 8 - 3 = 5 (nos quedan 5 conejos en la granja)

</div>

## Una resta directa de dos números

```python
# Ejecuta el código para que veas el resultado
10 - 4
```

## Si existen variables con valores numéricos, también se pueden restar. Usaremos las mismas variables que creamos en la suma

### - Ahora restaremos los valores

```python
conejos_blancos - conejos_negros
```

### - Valores de una variable - un valor numérico directo

```python
conejos_blancos - 3
```

### - Al igual que en la suma, puedes restar muchos valores al mismo tiempo. Sin embargo, a diferencia de la suma, en la resta el orden SÍ importa

```python
conejos_blancos - conejos_negros - 2
```

# ✖️ Multiplicación - Cuando los conejos se multiplican

La multiplicación nos ayuda a contar repeticiones. Por ejemplo, si tenemos 3 grupos de conejos y cada grupo tiene 4 conejos, ¿cuántos conejos tenemos en total?

**Símbolo:** *

**Ejemplo:** 3 * 4 = 12 (tenemos 12 conejos en total)

</div>

## Python permite multiplicar números de varias formas:

### - Una multiplicación directa de dos números

```python
# Ejecuta el código para que veas el resultado
4 * 3
```

## Si existen variables con valores numéricos, también se pueden multiplicar. Usaremos las mismas variables que creamos anteriormente

### - Ahora multiplicaremos los valores

```python
conejos_blancos * conejos_negros
```

### - Valores de una variable * un valor numérico directo

```python
conejos_blancos * 3
```

### - Al igual que las operaciones anteriores, puedes multiplicar muchos valores al mismo tiempo

```python
conejos_blancos * conejos_negros * 2
```

# ➗ División - Repartiendo entre varios

La división nos ayuda a repartir cosas. Por ejemplo, si tenemos 20 zanahorias y queremos repartirlas entre 4 conejos, ¿cuántas le tocan a cada uno?

**Símbolo:** /

**Ejemplo:** 20 / 4 = 5.0 (cada conejo recibe 5 zanahorias)

</div>

## Una división directa de dos números

```python
# Ejecuta el código para que veas el resultado
20 / 4
```

## También permite hacerlo entre variables y números directos -> lo omitiré para no saturarte, hasta este punto se entiendo de que va

<div style="border: 3px solid #4CAF50; padding: 15px; border-radius: 10px; background-color: #F1F8E9;">

# 🧮 División Entera - Solo la parte entera

La división entera nos da solo la parte entera del resultado, sin decimales. Es como repartir cosas y quedarse solo con las partes completas.

**Símbolo:** //

**Ejemplo:** 9 // 2 = 4 (porque 9 ÷ 2 = 4.5, pero solo tomamos el 4)

</div>

## División entera con números directos

```python
# División entera básica
9 // 2
```

## División entera con variables

```python
# División entera entre conejos blancos y negros
conejos_blancos // conejos_negros
```

## - También permite hacerlo entre variables y números directos -> lo omitiré para no saturarte

<div style="border: 3px solid #FF9800; padding: 15px; border-radius: 10px; background-color: #FFF3E0;">

# 🔢 Módulo - Lo que sobra

El módulo nos dice cuánto sobra después de una división. Es como repartir caramelos y ver cuántos quedan.

**Símbolo:** %

**Ejemplo:** 9 % 2 = 1 (porque 9 ÷ 2 = 4 y sobra 1)

</div>

## Módulo con números directos

```python
# Módulo básico - lo que sobra
9 % 2
```

## Módulo con variables

```python
# Lo que sobra al dividir conejos blancos entre negros
conejos_blancos % conejos_negros
```

<div style="border: 3px solid #9C27B0; padding: 15px; border-radius: 10px; background-color: #F3E5F5;">

# ⚡ Potencia - Multiplicar por sí mismo

La potencia multiplica un número por sí mismo varias veces. Es como magia exponencial.

**Símbolo:** **

**Ejemplo:** 2 ** 3 = 8 (porque 2 × 2 × 2 = 8)

</div>

## Potencia con números directos

```python
# Potencia básica - 2 elevado a la 3
2 ** 3
```

## Potencia con variables

```python
# Conejos negros elevado a conejos negros (¡magia exponencial!)
conejos_negros ** conejos_negros
```

## Combinando variable y número

```python
# Conejos blancos elevado al cuadrado
conejos_blancos ** 2
```

<h3>🧠 Recordatorio mnemotécnico:</h3>

<strong>🎪 "Por favor, Perdone Mi Querida Suma"</strong><br>
<em>Paréntesis → Potencias → Multiplicación/División → Suma/Resta</em>
</blockquote>

<h3>🎭 Ejemplos paso a paso:</h3>

<h4>📌 Ejemplo 1: Sin paréntesis</h4>
<pre style="background:#f3e5f5; padding:0.8em; border-radius:6px;">
expresion = 2 + 3 * 4
# Paso 1: 3 * 4 = 12     (multiplicación primero)
# Paso 2: 2 + 12 = 14    (luego suma)
# Resultado: 14
</pre>

<h4>📌 Ejemplo 2: Con paréntesis</h4>
<pre style="background:#e1f5fe; padding:0.8em; border-radius:6px;">
expresion = (2 + 3) * 4
# Paso 1: (2 + 3) = 5    (paréntesis primero)
# Paso 2: 5 * 4 = 20     (luego multiplicación)
# Resultado: 20
</pre>

<h4>📌 Ejemplo 3: Con potencias</h4>
<pre style="background:#fff3e0; padding:0.8em; border-radius:6px;">
expresion = 2 + 3 ** 2 * 4
# Paso 1: 3 ** 2 = 9     (potencia primero)
# Paso 2: 9 * 4 = 36     (multiplicación)
# Paso 3: 2 + 36 = 38    (suma al final)
# Resultado: 38
</pre>

<h3>⚠️ Casos especiales importantes:</h3>

<div style="background:#fff8e1; padding:1em; border-radius:6px; border-left:4px solid #ffc107;">

<h4>🔗 Operadores del mismo nivel:</h4>
<p>Cuando hay operadores de la misma prioridad, se evalúan <strong>de izquierda a derecha</strong>:</p>

<pre style="background:#ffffff; padding:0.6em; border-radius:4px; margin:0.5em 0;">
# Suma y resta (mismo nivel)
resultado = 10 - 3 + 2    # (10 - 3) + 2 = 7 + 2 = 9

# Multiplicación y división (mismo nivel)  
resultado = 12 / 3 * 2    # (12 / 3) * 2 = 4.0 * 2 = 8.0
</pre>

<h4>⚡ Excepción: Potencias van de derecha a izquierda</h4>
<pre style="background:#ffffff; padding:0.6em; border-radius:4px; margin:0.5em 0;">
resultado = 2 ** 3 ** 2   # 2 ** (3 ** 2) = 2 ** 9 = 512
# ¡NO es (2 ** 3) ** 2 = 8 ** 2 = 64!
</pre>

</div>

<h3>💡 Consejos prácticos para evitar confusiones:</h3>

<ul style="background:#e8f5e9; padding:1em; border-radius:6px;">
  <li><strong>🔐 Usa paréntesis liberalmente:</strong> <code>(a + b) * c</code> es más claro que <code>a + b * c</code></li>
  <li><strong>📝 Divide expresiones complejas:</strong> Mejor varias líneas simples que una línea complicada</li>
  <li><strong>🧪 Prueba paso a paso:</strong> Si no estás seguro, evalúa cada parte por separado</li>
  <li><strong>📖 Lee de izquierda a derecha:</strong> Excepto potencias que van de derecha a izquierda</li>
</ul>

<hr style="border: none; border-top: 1px solid #81c784;" />

<p><em style="color:#2e7d32;"><strong>🎯 Regla de oro:</strong> Si tienes dudas sobre el orden, ¡usa paréntesis! Es mejor ser explícito que confundirse. Los paréntesis nunca sobran en programación 🧙‍♂️</em></p>

</div>

```python
# 📏 Ejemplo 1: El poder de los paréntesis

print("🔍 === SIN PARÉNTESIS VS CON PARÉNTESIS ===")

# Sin paréntesis - sigue la jerarquía natural
expresion1 = 2 + 3 * 4
print(f"Sin paréntesis: 2 + 3 * 4 = {expresion1}")
print("   Se evalúa: 2 + (3 * 4) = 2 + 12 = 14")

# Con paréntesis - fuerza el orden
expresion2 = (2 + 3) * 4  
print(f"Con paréntesis: (2 + 3) * 4 = {expresion2}")
print("   Se evalúa: (2 + 3) * 4 = 5 * 4 = 20")

print(f"\n¿Son iguales? {expresion1 == expresion2}")
print("💡 ¡Los paréntesis cambian completamente el resultado!")
```

```python
# ⚡ Ejemplo 2: Las potencias van primero

print("⚡ === LAS POTENCIAS TIENEN PRIORIDAD ===")

expresion3 = 2 + 3 ** 2 * 4
print("Expresión: 2 + 3 ** 2 * 4")
print(f"   Paso 1: 3 ** 2 = {3 ** 2}")
print(f"   Paso 2: {3 ** 2} * 4 = {3 ** 2 * 4}")
print(f"   Paso 3: 2 + {3 ** 2 * 4} = {expresion3}")

print(f"\nResultado final: {expresion3}")
print("💡 Orden: Potencia → Multiplicación → Suma")
```

```python
# ➡️ Ejemplo 3: Mismo nivel = izquierda a derecha

print("➡️ === MISMO NIVEL: IZQUIERDA A DERECHA ===")

# Multiplicación y división (mismo nivel)
expresion4 = 20 / 4 * 3
expresion5 = 20 / (4 * 3)

print("Expresión: 20 / 4 * 3")
print(f"   Se evalúa: (20 / 4) * 3 = {20 / 4} * 3 = {expresion4}")
print(f"   NO es: 20 / (4 * 3) = 20 / {4 * 3} = {expresion5}")

print(f"\n💡 Mismo nivel → de izquierda a derecha")

# 🔄 Excepción: potencias van de derecha a izquierda  
print("\n🔄 === EXCEPCIÓN: POTENCIAS (DERECHA A IZQUIERDA) ===")

expresion6 = 2 ** 3 ** 2
expresion7 = (2 ** 3) ** 2

print("Expresión: 2 ** 3 ** 2")
print(f"   Se evalúa: 2 ** (3 ** 2) = 2 ** {3 ** 2} = {expresion6}")
print(f"   NO es: (2 ** 3) ** 2 = {2 ** 3} ** 2 = {expresion7}")

print(f"\n⚡ Potencias van de derecha a izquierda (única excepción)")
```

```python
# 🏪 Ejemplo práctico: calculadora confusa vs clara

print("🏪 === CALCULADORA DE TIENDA: CONFUSA VS CLARA ===")

# Variables de una tienda mágica
precio_base = 50
descuento_porcentaje = 20
impuesto_porcentaje = 10
cantidad = 3

# ❌ Forma confusa (una sola línea muy larga)
total_confuso = precio_base * cantidad - precio_base * cantidad * descuento_porcentaje / 100 + precio_base * cantidad * impuesto_porcentaje / 100

print("💀 Cálculo confuso (una sola línea):")
print(f"   Total = {total_confuso}")

# ✅ Forma clara (paso a paso)
print(f"\n✨ Cálculo claro (paso a paso):")
subtotal = precio_base * cantidad
print(f"   1. Subtotal: {precio_base} × {cantidad} = ${subtotal}")

descuento = subtotal * (descuento_porcentaje / 100)
print(f"   2. Descuento: ${subtotal} × {descuento_porcentaje}% = ${descuento}")

despues_descuento = subtotal - descuento
print(f"   3. Después del descuento: ${subtotal} - ${descuento} = ${despues_descuento}")

impuesto = despues_descuento * (impuesto_porcentaje / 100)
print(f"   4. Impuesto: ${despues_descuento} × {impuesto_porcentaje}% = ${impuesto}")

total_claro = despues_descuento + impuesto
print(f"   5. Total final: ${despues_descuento} + ${impuesto} = ${total_claro}")

print(f"\n🔍 ¿Los resultados coinciden? {abs(total_confuso - total_claro) < 0.01}")
print("💡 ¡Es mejor ser claro que confuso!")
```

# **Operadores lógicos**

<div style="border-left: 6px solid #00897b; padding: 1.5em; background: #e0f2f1; border-radius: 12px; font-family: 'Segoe UI', sans-serif; color: #004d40; line-height: 1.6;">

<h2 style="color:#00796b;">🤯 No es lo mismo los aritméticos</h2>

<p>Ahora que ya conoces las variables <code>True</code> y <code>False</code>, es momento de combinarlas. ¡Como hacer un hechizo con más de un ingrediente!</p>

<h3>🐰 1. <code>and</code> — "Y además..."</h3>
<p>Solo devuelve <code>True</code> si <strong>todo es verdadero</strong>.</p>

<pre style="background:#c8e6c9; padding:0.8em; border-radius:6px;">
tiene_varita = True
sabe_hechizo = True
print(tiene_varita and sabe_hechizo)   # ✅ True
</pre>

<p>Pero si uno falla...</p>

<pre style="background:#ffcdd2; padding:0.8em; border-radius:6px;">
tiene_varita = True
sabe_hechizo = False
print(tiene_varita and sabe_hechizo)   # ❌ False
</pre>

<h3>🦄 2. <code>or</code> — "Con que uno funcione..."</h3>
<p>Devuelve <code>True</code> si <strong>al menos uno</strong> es verdadero.</p>

<pre style="background:#fff9c4; padding:0.8em; border-radius:6px;">
tiene_capa = False
tiene_escoba = True
print(tiene_capa or tiene_escoba)   # ✅ True
</pre>

<h3>🙃 3. <code>not</code> — "Lo contrario, por favor"</h3>
<p>Cambia <code>True</code> a <code>False</code> y viceversa. ¡Es un hechizo de reversa!</p>

<pre style="background:#fce4ec; padding:0.8em; border-radius:6px;">
esta_lloviendo = True
print(not esta_lloviendo)   # Resultado: False
</pre>

<h3>📌 Consejo mágico:</h3>
<ul>
  <li><code>and</code> necesita que todo sea cierto 🧠</li>
  <li><code>or</code> se conforma con que algo sea cierto ✨</li>
  <li><code>not</code> voltea el resultado como si fuera una media 🧦</li>
</ul>

```python
# 🔍 Practicando comparaciones: ¡conviértete en un juez mágico!

print("⚖️ === COMPARACIONES NUMÉRICAS ===")

# Variables de ejemplo
conejos_blancos = 8
conejos_marrones = 5
conejos_grises = 8

# Comparaciones básicas
print("¿Hay más conejos blancos que marrones?", conejos_blancos > conejos_marrones)
print("¿Hay menos conejos marrones que blancos?", conejos_marrones < conejos_blancos)
print("¿Los conejos blancos y grises son iguales?", conejos_blancos == conejos_grises)
print("¿Los conejos blancos y marrones son diferentes?", conejos_blancos != conejos_marrones)

# Comparaciones con >= y <=
edad_minima_escuela = 5
edad_conejo = 5
print(f"¿Puede ir a la escuela? (edad >= {edad_minima_escuela})", edad_conejo >= edad_minima_escuela)
```

```python
print("\n📝 === COMPARACIONES DE TEXTO ===")

# Variables de texto
nombre1 = "Canelita"
nombre2 = "Canelita"
nombre3 = "Canelita"
hechizo = "Abracadabra"

print(f"¿'{nombre1}' es igual a '{nombre2}'?", nombre1 == nombre2)
print(f"¿'{nombre1}' es igual a '{nombre3}'?", nombre1 == nombre3)
print(f"¿'{nombre1}' es diferente de '{nombre2}'?", nombre1 != nombre2)

# Truco para ignorar mayúsculas/minúsculas
print(f"¿Son iguales ignorando mayúsculas?", nombre1.lower() == nombre2.lower())

# Comparaciones con longitud de texto
print(f"¿El hechizo '{hechizo}' tiene más de 10 caracteres?", len(hechizo) > 10)

print("\n🧮 === COMBINANDO COMPARACIONES CON OPERADORES LÓGICOS ===")

# Variables para ejemplos
tiene_varita = True
edad = 7
poder_magico = 85

# Usando and
es_mago_completo = tiene_varita and edad >= 5 and poder_magico > 80
print("¿Es un mago completo?", es_mago_completo)

# Usando or
puede_entrar_castillo = tiene_varita or poder_magico > 90
print("¿Puede entrar al castillo?", puede_entrar_castillo)
```

```python
print("\n🎲 === DESAFÍOS DE COMPARACIÓN ===")
# ¿Puedes predecir qué resultará cada uno?

a, b, c = 10, 5, 10
print("Desafío 1:", a == c and b < a)
print("Desafío 2:", a != b or b == c)
print("Desafío 3:", not (a < b))
print("Desafío 4:", len("Python") == len("Conejo"))
print("Desafío 5:", "abc" < "xyz")  # ¡Las letras también se pueden comparar!
```
