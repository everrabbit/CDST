# 🗂️ ¿Qué es un DataFrame?

Un **DataFrame** es como una **hoja de Excel mágica** dentro de Python. Es una tabla con filas y columnas donde puedes almacenar y manipular datos de manera súper eficiente.

**💭 Piénsalo como:**
- Una tabla de Excel, pero con superpoderes 🦸‍♀️
- Una base de datos ligera y fácil de usar
- Una grilla donde cada columna puede tener diferentes tipos de datos

## 🏗️ ¿De dónde vienen los DataFrames?

Los DataFrames se pueden crear desde muchas fuentes:

| 📁 Fuente | 📝 Descripción | 🎯 Ejemplo |
|-----------|----------------|------------|
| **Diccionarios** | Datos creados a mano | `{'nombre': ['Ana', 'Luis'], 'edad': [25, 30]}` |
| **Archivos CSV** | Hojas de cálculo | `pd.read_csv('datos.csv')` |
| **Archivos Excel** | Workbooks de Excel | `pd.read_excel('archivo.xlsx')` |
| **Bases de datos** | SQL, SQLite, etc. | `pd.read_sql()` |
| **APIs web** | Datos de internet | `pd.read_json()` |
| **Listas** | Arrays de Python | `pd.DataFrame(lista_de_listas)` |

</div>

# 🎨 ¡Creemos nuestro primer DataFrame!

Vamos a crear una tabla con información de una granja de conejos mágicos. Usaremos un **diccionario** como base.

**🗂️ Estructura del diccionario:**
```python
datos = {
    'columna1': [valor1, valor2, valor3],
    'columna2': [valor1, valor2, valor3]
}
```

**🎯 Resultado:** Una tabla hermosa y organizada donde cada clave del diccionario se convierte en una columna.

</div>

````python
# 🐰 Creando nuestra granja de conejos mágicos (¡sí, mis primos!)

# Primero importamos pandas
import pandas as pd

print("🐰✨ === CREANDO NUESTRO PRIMER DATAFRAME ===\n")
print("🎩 *El conejo aprendiz agita su varita... '¡Abracadabra DataFramus!'*")

# Creamos un diccionario con datos de conejos
# 🐰 *'Estos son mis primos de la granja mágica' - dice el conejo con orgullo*
datos_conejos = {
    'nombre': ['Pelusa', 'Saltarín', 'Copito', 'Brillante', 'Veloz'],
    'color': ['Blanco', 'Marrón', 'Blanco', 'Gris', 'Negro'],
    'edad': [2, 4, 1, 3, 2],
    'peso': [1.5, 2.1, 1.2, 1.8, 1.6],
    'es_magico': [True, False, True, True, False]
}

# Convertimos el diccionario en un DataFrame
df_conejos = pd.DataFrame(datos_conejos)

print("✨ ¡POOF! DataFrame creado exitosamente!")
print(f"📊 Dimensiones: {df_conejos.shape[0]} conejos x {df_conejos.shape[1]} características")
print("🐰 *'¡Son más organizados que mi madriguera!' - exclama el conejo*\n")

# Mostramos nuestro DataFrame
print("🗂️ Nuestra granja de conejos (versión 2.0 de mi familia):")
display(df_conejos)
print("\n🥕 *El conejo susurra: 'Saltarín es mi primo favorito, aunque no sea mágico...'*")
````

# 🔍 Explorando nuestro DataFrame

Ahora que tenemos nuestro DataFrame, vamos a explorarlo como verdaderos detectives de datos.

## 🧙‍♂️ Comandos mágicos básicos:

| 🪄 Comando | 📝 Descripción | 🎯 Ejemplo |
|------------|----------------|------------|
| `df.head()` | Muestra las primeras 5 filas | `df_conejos.head()` |
| `df.tail()` | Muestra las últimas 5 filas | `df_conejos.tail()` |
| `df.info()` | Información general del DataFrame | `df_conejos.info()` |
| `df.describe()` | Estadísticas de columnas numéricas | `df_conejos.describe()` |
| `df.shape` | Dimensiones (filas, columnas) | `df_conejos.shape` |
| `df.columns` | Nombres de las columnas | `df_conejos.columns` |

````python
# 🕵️ Explorando nuestra granja como detectives de datos (¡con lupa mágica!)

print("🔍✨ === EXPLORANDO NUESTRO DATAFRAME ===\n")
print("🐰 *'¡Vamos a investigar a mis primos como verdaderos detectives!' - dice el conejo*")

# 1️⃣ Información general
print("📋 INFORMACIÓN GENERAL:")
print(f"   • Dimensiones: {df_conejos.shape} (más ordenado que mi madriguera)")
print(f"   • Columnas: {list(df_conejos.columns)}")
print(f"   • Total de conejos: {len(df_conejos)} (una familia pequeña pero poderosa)")

print("\n" + "="*60)

# 2️⃣ Primeras filas
print("\n👀 PRIMERAS 3 FILAS (los conejos más famosos):")
display(df_conejos.head(3))
print("🐰 *'Pelusa siempre quiere ser la primera' - suspira el conejo*")

print("\n" + "="*60)

# 3️⃣ Información detallada de tipos de datos
print("\n📊 TIPOS DE DATOS (o 'de qué están hechos mis primos'):")
print(df_conejos.dtypes)
print("🔮 *'Los datos mágicos son booleanos, ¡obvio!' - dice el conejo orgulloso*")

print("\n" + "="*60)

# 4️⃣ Estadísticas de columnas numéricas
print("\n📈 ESTADÍSTICAS NUMÉRICAS (¡las matemáticas mágicas!):")
display(df_conejos.describe())
print("🥕 *'Copito es el más pequeñito, ¡pero no subestimes su magia!' - advierte el conejo*")
````

````python
# 🎯 Operaciones avanzadas con DataFrames (¡las travesuras del conejo!)

print("🚀✨ === MANIPULACIÓN AVANZADA DE DATAFRAMES ===\n")
print("🐰 *'¡Ahora vamos a filtrar a mi familia como si fuéramos paparazzi!' - dice el conejo travieso*")

# 1️⃣ Filtrado de datos
print("🔍 FILTRADO DE DATOS (separando a los especiales):")
print("-" * 50)

# Filtrar conejos mágicos
conejos_magicos = df_conejos[df_conejos['es_magico'] == True]
print("✨ Conejos mágicos (la élite de la familia):")
display(conejos_magicos)
print("🎩 *'¡Estos son mis favoritos!' - guiña el ojo el conejo*")

print("\n" + "="*60)

# Filtrar por edad
print("\n👶 Conejos jóvenes (los bebés rebeldes menores a 3 años):")
conejos_jovenes = df_conejos[df_conejos['edad'] < 3]
display(conejos_jovenes)
print("🍼 *'Copito todavía usa pañal mágico' - ríe el conejo*")

print("\n" + "="*60)

# 2️⃣ Múltiples filtros
print("\n🎯 FILTROS COMBINADOS (¡el combo perfecto!):")
print("-" * 50)

# Conejos blancos Y mágicos
conejos_especiales = df_conejos[(df_conejos['color'] == 'Blanco') & 
                                (df_conejos['es_magico'] == True)]
print("🤍✨ Conejos blancos y mágicos (la crème de la crème):")
display(conejos_especiales)
print("👑 *'Estos son los VIP de la familia' - dice el conejo con orgullo*")

print("\n" + "="*60)

# 3️⃣ Selección de columnas específicas
print("\n📊 SELECCIÓN DE COLUMNAS (como elegir qué chisme contar):")
print("-" * 50)

# Solo nombres y edades
nombres_edades = df_conejos[['nombre', 'edad']]
print("📝 Solo nombres y edades (lo básico para el chismorreo):")
display(nombres_edades)
print("🗣️ *'Con esto ya puedes hacer buenos chismes en la madriguera' - susurra el conejo*")

print("\n" + "="*60)

# 4️⃣ Ordenamiento
print("\n🔤 ORDENAMIENTO (poniendo orden en el caos):")
print("-" * 50)

# Ordenar por edad (de menor a mayor)
por_edad = df_conejos.sort_values('edad')
print("📈 Ordenados por edad (del bebé al veterano):")
display(por_edad)

print("\n📉 Ordenados por peso (de los gorditos a los flaquitos):")
por_peso = df_conejos.sort_values('peso', ascending=False)
display(por_peso)
print("🥕 *'Saltarín come demasiadas zanahorias' - comenta el conejo*")

print("\n🎉 ¡Ya dominas las operaciones básicas con DataFrames!")
print("🐰 *'¡Eres oficialmente un mago de los datos!' - aplaude el conejo*")
print("🎩✨ *'Ahora puedes manipular datos mejor que yo mis trucos de magia'*")
````
